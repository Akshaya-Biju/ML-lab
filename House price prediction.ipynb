```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Reading the data file
df = pd.read_csv('/content/housing.csv')
#print the datframe
print(df)
#printing the shape
print(df.shape)
#printing top 10 and bottom 10 rows
print(df.head(10))
print(df.tail(10))
#printing the columns
print(df.columns)
#print a feature value
print(df['Avg. Area Number of Rooms'])
# Selecting one input feature and target feature (dependent) for 150 rows only
X = df['Avg. Area Number of Rooms'].head(150).values.reshape(-1,1)
y = df['Price'].head(150).values.reshape(-1, 1)

#visualizing these values
plt.scatter(X, y)
plt.xlabel("Avg. Area Number of Rooms")
plt.ylabel("Price")
plt.title("Scatter Plot of Avg. Area Number of Rooms vs. Price")
plt.show()

#printing the statistical summary
print(df.describe())
```
